```python
import torch
import transformers
import peft
import torchao

print("PyTorch :", torch.__version__)
print("Transformers :", transformers.__version__)
print("PEFT :", peft.__version__)
print("CUDA :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU :", torch.cuda.get_device_name(0))

print("torchao :", torchao.__version__)
```

```python
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
)

from peft import PeftModel

MODEL_ID = "bivariant/griot-mt"

print("🔄 Chargement du tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)

print("🔄 Chargement du modèle...")

device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if device == "cuda" else torch.float32

base_model = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_ID,
    torch_dtype=dtype,
).to(device)

print("🔄 Chargement de l'adaptateur Baoulé...")

model = PeftModel.from_pretrained(
    base_model,
    MODEL_ID,
    subfolder="adapters/bci",
)

model.eval()

print("✅ Griot-MT Baoulé prêt !")
print("GPU :", device)
```

```python
LANGUAGE_TOKENS = {
    "bci": "bci_Latn",
}

FRENCH_TOKEN = "fra_Latn"

def traduire_baoule(texte):
    tokenizer.src_lang = LANGUAGE_TOKENS["bci"]

    inputs = tokenizer(
        texte,
        return_tensors="pt",
        truncation=True
    ).to(device)

    with torch.inference_mode():
        output_ids = model.generate(
            **inputs,
            forced_bos_token_id=tokenizer.convert_tokens_to_ids(
                FRENCH_TOKEN
            ),
            num_beams=4,
            max_length=257,
            repetition_penalty=1.3,
            no_repeat_ngram_size=3
        )

    return tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )


texte = "I sin’n, ɔ mɛnnin i wun’n wie. Yɛ i kusu dili ɔ."

print("🇨🇮 Baoulé :", texte)
print("🇫🇷 Traduction :", traduire_baoule(texte))
```

```python
# charger NLLB Baoulé v1
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_ID = "GaindeNdiaye/nllb-baoule-v1"

print("🔄 Chargement du tokenizer...")

tokenizer_nllb = AutoTokenizer.from_pretrained(MODEL_ID)

print("🔄 Chargement du modèle...")

device_nllb = "cuda" if torch.cuda.is_available() else "cpu"
dtype_nllb = torch.float16 if device_nllb == "cuda" else torch.float32

model_nllb = AutoModelForSeq2SeqLM.from_pretrained(
    MODEL_ID,
    dtype=dtype_nllb
).to(device_nllb)

model_nllb.eval()

print("✅ NLLB Baoulé chargé !")
print("GPU :", device_nllb)
```

```python
# tester la même phrase pour voir qui est le meilleur modele
texte_baoule = "I sin’n, ɔ mɛnnin i wun’n wie. Yɛ i kusu dili ɔ."

print("🇨🇮 Baoulé :", texte_baoule)

tokenizer_nllb.src_lang = "bci_Latn"

inputs = tokenizer_nllb(
    texte_baoule,
    return_tensors="pt",
    truncation=True
).to(device_nllb)

with torch.inference_mode():
    output = model_nllb.generate(
        **inputs,
        forced_bos_token_id=tokenizer_nllb.convert_tokens_to_ids("fra_Latn"),
        max_new_tokens=128,
        num_beams=4
    )

traduction_nllb = tokenizer_nllb.decode(
    output[0],
    skip_special_tokens=True
)

print("🇫🇷 Traduction NLLB :", traduction_nllb)
```

```python
print("Modèle Griot-MT chargé :", "model" in globals())
print("Tokenizer chargé :", "tokenizer" in globals())
print("GPU disponible :", torch.cuda.is_available())
```

```python
# Tester si le modele Griot-MT peut traduire du francais au baoulé
# 🇫🇷 Phrase à traduire : modifie cette ligne
phrase_francaise = "J’ai appris qu’il n’y a rien de mieux pour [les humains] que de se réjouir et de faire le bien durant [leur] vie, a écrit Salomon ; et aussi que tout hommemange, oui qu’il boive et qu’il voie le bien pour tout son dur travail."

# Langue source : français
tokenizer.src_lang = "fra_Latn"

# Préparer le texte
inputs = tokenizer(
    phrase_francaise,
    return_tensors="pt",
    truncation=True
).to(device)

# Générer la traduction en Baoulé
with torch.inference_mode():
    output_ids = model.generate(
        **inputs,
        forced_bos_token_id=tokenizer.convert_tokens_to_ids("bci_Latn"),
        num_beams=4,
        max_new_tokens=128,
        repetition_penalty=1.2
    )

# Afficher le résultat
traduction_baoule = tokenizer.decode(
    output_ids[0],
    skip_special_tokens=True
)

print("🇫🇷 Français :", phrase_francaise)
print("🇨🇮 Baoulé :", traduction_baoule)
```
